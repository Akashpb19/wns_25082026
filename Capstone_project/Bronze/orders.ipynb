```python
order_df = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option(
        "cloudFiles.schemaLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/autoloader_metadata/orders/"
    )
    .option("header", "true")
    .load(
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/datasets/sql_server/incremental/orders/"
    )
)


query = (
    order_df.writeStream
    .format("delta")
    .outputMode("append")
    .option(
        "checkpointLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/checkpoints/orders/"
    )
    .trigger(availableNow=True)
    .toTable("ecommerce.bronze.orders")
)
```
