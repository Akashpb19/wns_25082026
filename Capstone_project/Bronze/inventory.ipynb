```python
inventory_df = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option(
        "cloudFiles.schemaLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/autoloader_metadata/inventory/"
    )
    .option("header", "true")
    .load(
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/datasets/adls/incremental/inventory/"
    )
)


query = (
    inventory_df.writeStream
    .format("delta")
    .outputMode("append")
    .option(
        "checkpointLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/checkpoints/inventory/"
    )
    .trigger(availableNow=True)
    .toTable("ecommerce.bronze.inventory")
)
```
