```python
order_items_df = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option(
        "cloudFiles.schemaLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/autoloader_metadata/order_items/"
    )
    .option("header", "true")
    .load(
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/datasets/sql_server/incremental/order_items/"
    )
)


query = (
    order_items_df.writeStream
    .format("delta")
    .outputMode("append")
    .option(
        "checkpointLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/checkpoints/order_items/"
    )
    .trigger(availableNow=True)
    .toTable("ecommerce.bronze.order_items")
)
```
