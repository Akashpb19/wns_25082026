```python
products_df = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option(
        "cloudFiles.schemaLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/autoloader_metadata/products/"
    )
    .option("header", "true")
    .load(
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/datasets/adls/incremental/products/"
    )
)


query = (
    products_df.writeStream
    .format("delta")
    .outputMode("append")
    .option(
        "checkpointLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/checkpoints/products/"
    )
    .trigger(availableNow=True)
    .toTable("ecommerce.bronze.products")
)
```
