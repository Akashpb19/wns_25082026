## Load any incremental data to work with 

```python
customer_df = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option(
        "cloudFiles.schemaLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/autoloader_metadata/customers/"
    )
    .option("header", "true")
    .load(
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/datasets/sql_server/incremental/customers/"
    )
)
```

## Table Drop for the first time for demonstration purposes

We wll drop the table if exists for the first time. We are not going to use it again and infact we will comment it for future usage 

```python
# %sql
# DROP TABLE IF EXISTS ECOMMERCE.BRONZE.CUSTOMERS;
```

## Writing data to a given table 
We will now write the data to a given table as per the requirement 

```python
query = (
    customer_df.writeStream
    .format("delta")
    .outputMode("append")
    .option(
        "checkpointLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/checkpoints/customers/"
    )
    .trigger(availableNow=True)
    .toTable("ecommerce.bronze.customers")
)
```
