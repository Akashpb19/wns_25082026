```python
clickstream_df = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option(
        "cloudFiles.schemaLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/autoloader_metadata/clickstream/"
    )
    .load(
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/datasets/adls/incremental/clickstream/"
    )
)

query = (
    clickstream_df.writeStream
    .format("delta")
    .outputMode("append")
    .option(
        "checkpointLocation",
        "/Volumes/capstone_project_wns24082026/capstone/datasets_assigned/checkpoints/clickstream/"
    )
    .trigger(availableNow=True)
    .toTable("ecommerce.bronze.clickstream")
)
```
